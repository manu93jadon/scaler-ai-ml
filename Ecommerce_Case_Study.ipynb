# 1 ) Identify the months with the highest and lowest acquisition count. What strategies could be implemented to address the fluctuations and ensure consistent growth throughout the year

To get the monthly acquisitions, we need to find the first time each customer ever made a purchase. If a customer bought something in January, March, and July, their acquisition month is January.

```python
first_purchase = Online_Sales.groupby('CustomerID')['Transaction_Date'].min().reset_index()
first_purchase.columns = ['CustomerID', 'Acquisition_Date']
first_purchase['Acquisition_Month_Name'] = first_purchase['Acquisition_Date'].dt.strftime('%B')
first_purchase['Acquisition_Month_Num'] = first_purchase['Acquisition_Date'].dt.month
monthly_acquisitions = first_purchase.groupby(['Acquisition_Month_Num', 'Acquisition_Month_Name']).size().reset_index(name='Acquisition_Count')
monthly_acquisitions = monthly_acquisitions.drop(columns=['Acquisition_Month_Num'])
monthly_acquisitions = monthly_acquisitions.sort_values(by='Acquisition_Count', ascending=False)
print(monthly_acquisitions)
```

Above Table shows that January March and April recieves highest customer acquisition. Whereas October, September and November the minimum. We will now try to figure out what could be the reason.

# 3 ) Identify periods with the strongest and weakest retention rates. What strategies could be implemented to improve retention during weaker months?

## Let us establish the definition of Retention Rate which we are going to follow. When Googled, we can see - $$Retention\ Rate\ (\%) = \left( \frac{\text{Number of customers from a cohort who bought in Month } X}{\text{Total unique customers who started in that Cohort}} \right) \times 100$$

We will follow the same strategy through out the EDA

### Calculate the Retention Matrix using Online_Sales Dataset

```python
Online_Sales['Transaction_Month'] = Online_Sales['Transaction_Date'].dt.to_period('M')
Online_Sales['Cohort_Month'] = Online_Sales.groupby('CustomerID')['Transaction_Month'].transform('min')

def get_month_difference(earliest_date, current_date):
    return (current_date.year - earliest_date.year) * 12 + (current_date.month - earliest_date.month)

Online_Sales['Cohort_Index'] = Online_Sales.apply(lambda row: get_month_difference(row['Cohort_Month'], row['Transaction_Month']), axis=1)
```

### Constructing the Retention Matrix

```python
# Group by Cohort_Month and Cohort_Index to count unique customers
cohort_counts = Online_Sales.groupby(['Cohort_Month', 'Cohort_Index']).agg(N_Customers=('CustomerID', 'nunique')).reset_index()
display(cohort_counts.head())
```

```python
# Calculate the initial cohort size (number of unique customers in the Cohort_Month)
cohort_sizes = cohort_counts[cohort_counts['Cohort_Index'] == 0][['Cohort_Month', 'N_Customers']]
cohort_sizes.rename(columns={'N_Customers': 'Cohort_Size'}, inplace=True)
display(cohort_sizes.head())
```

```python
# Merge cohort counts with cohort sizes to calculate retention rate
retention_matrix_df = pd.merge(cohort_counts, cohort_sizes, on='Cohort_Month', how='left')
retention_matrix_df['Retention_Percentage'] = (retention_matrix_df['N_Customers'] / retention_matrix_df['Cohort_Size']) * 100
display(retention_matrix_df.head())
```

```python
# Pivot the table to create the retention matrix
retention_pivot = retention_matrix_df.pivot_table(index='Cohort_Month', columns='Cohort_Index', values='Retention_Percentage')
display(retention_pivot)
```

### Visualizing the Retention Matrix as a Heatmap

```python
plt.figure(figsize=(12, 8))
sns.heatmap(retention_pivot, annot=True, fmt='.1f', cmap='Blues', vmin=0, vmax=100)
plt.title('Customer Retention Matrix')
plt.xlabel('Cohort Index (Months Since First Purchase)')
plt.ylabel('Cohort Month')
plt.show()
```

##Conclusion:
> Strong Periods:



1.   Immediate Retention Rate (0,1 Index Cohort,means they are returning immediately) is high in June and July
2.   Long Term Retention Rate: Jan and Feb, there are good numbers even after Cohort Index is > 4.




> Weak Periods:


1.   Immediate Drop - Jan, Feb and October though for Jan it makes sense as it saw very high number of acquisition to begin with.
2.   Long Term Drop - September and October - Their immediate and long term both are bad.

# 4 ) Analyze customer behavior during high-retention months and suggest ways to replicate this success throughout the year.

##As seen above , High Retention Months are Jan, Feb (Long term) and June , July ( Immediate ) .

Too deep dive further let us check Product Categories sold the most in high Retention Months.

```python
Online_Sales['Month'] = Online_Sales['Transaction_Date'].dt.month_name()

high_retention_months = ['January', 'February', 'June', 'July']
high_retention_sales = Online_Sales[Online_Sales['Month'].isin(high_retention_months)].copy()

# Calculate Total Price for each transaction
high_retention_sales['Total_Price'] = high_retention_sales['Quantity'] * high_retention_sales['Avg_Price']

print("### Product Category Performance in High-Retention Months")
product_category_sales = high_retention_sales.groupby(['Month', 'Product_Category']).agg(
    Total_Revenue=('Total_Price', 'sum'),
    Unique_Customers=('CustomerID', 'nunique')
).reset_index()

display(product_category_sales.sort_values(by=['Month', 'Total_Revenue'], ascending=[True, False]).head(10))
```

## We can target the customers showing High Retention in Jan Feb by offering the most discount on the product categories which got sold most like Nest-USA , Apparel and Office during the low retention periods. Strategic Discounting is certainly needed.

```python
low_retention_months = ['September', 'October']
low_retention_sales = Online_Sales[Online_Sales['Month'].isin(low_retention_months)].copy()

# Calculate Total Price for each transaction
low_retention_sales['Total_Price'] = low_retention_sales['Quantity'] * low_retention_sales['Avg_Price']

print("### Product Category Performance in Low-Retention Months")
product_category_sales_low_retention = low_retention_sales.groupby(['Month', 'Product_Category']).agg(
    Total_Revenue=('Total_Price', 'sum'),
    Unique_Customers=('CustomerID', 'nunique')
).reset_index()

display(product_category_sales_low_retention.sort_values(by=['Month', 'Total_Revenue'], ascending=[True, False]).head(10))
```

# 5 ) Compare the revenue generated by new and existing customers month-over-month. What does this trend suggest about the balance between acquisition and retention efforts?

```python
# Ensure 'Total_Price' is calculated in the main Online_Sales DataFrame
Online_Sales['Total_Price'] = Online_Sales['Quantity'] * Online_Sales['Avg_Price']

# Identify sales from new customers (Cohort_Index == 0)
new_customer_sales = Online_Sales[Online_Sales['Cohort_Index'] == 0].copy()

# Identify sales from existing customers (Cohort_Index > 0)
existing_customer_sales = Online_Sales[Online_Sales['Cohort_Index'] > 0].copy()

# Calculate monthly revenue for new customers
new_customer_revenue = new_customer_sales.groupby('Transaction_Month')['Total_Price'].sum().reset_index()
new_customer_revenue.rename(columns={'Total_Price': 'New_Customer_Revenue'}, inplace=True)

# Calculate monthly revenue for existing customers
existing_customer_revenue = existing_customer_sales.groupby('Transaction_Month')['Total_Price'].sum().reset_index()
existing_customer_revenue.rename(columns={'Total_Price': 'Existing_Customer_Revenue'}, inplace=True)

# Merge the revenues for comparison
monthly_revenue_comparison = pd.merge(new_customer_revenue, existing_customer_revenue, on='Transaction_Month', how='outer').fillna(0)

# Convert Transaction_Month to datetime for sorting and plotting
monthly_revenue_comparison['Month'] = monthly_revenue_comparison['Transaction_Month'].dt.to_timestamp().dt.strftime('%B')

# Define the desired order of months
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
monthly_revenue_comparison['Month'] = pd.Categorical(monthly_revenue_comparison['Month'], categories=month_order, ordered=True)
monthly_revenue_comparison = monthly_revenue_comparison.sort_values('Month')

display(monthly_revenue_comparison[['Month', 'New_Customer_Revenue', 'Existing_Customer_Revenue']])
```

```python
plt.figure(figsize=(14, 7))

sns.lineplot(x='Month', y='New_Customer_Revenue', data=monthly_revenue_comparison, marker='o', label='New Customer Revenue')
sns.lineplot(x='Month', y='Existing_Customer_Revenue', data=monthly_revenue_comparison, marker='o', label='Existing Customer Revenue')

plt.title('Monthly Revenue Comparison: New vs. Existing Customers')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```
