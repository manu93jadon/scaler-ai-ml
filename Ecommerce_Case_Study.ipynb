# 1 ) Identify the months with the highest and lowest acquisition count. What strategies could be implemented to address the fluctuations and ensure consistent growth throughout the year

To get the monthly acquisitions, we need to find the first time each customer ever made a purchase. If a customer bought something in January, March, and July, their acquisition month is January.

```python
first_purchase = Online_Sales.groupby('CustomerID')['Transaction_Date'].min().reset_index()
first_purchase.columns = ['CustomerID', 'Acquisition_Date']
first_purchase['Acquisition_Month_Name'] = first_purchase['Acquisition_Date'].dt.strftime('%B')
first_purchase['Acquisition_Month_Num'] = first_purchase['Acquisition_Date'].dt.month
monthly_acquisitions = first_purchase.groupby(['Acquisition_Month_Num', 'Acquisition_Month_Name']).size().reset_index(name='Acquisition_Count')
monthly_acquisitions = monthly_acquisitions.drop(columns=['Acquisition_Month_Num'])
monthly_acquisitions = monthly_acquisitions.sort_values(by='Acquisition_Count', ascending=False)
print(monthly_acquisitions)
```

Above Table shows that January March and April recieves highest customer acquisition. Whereas October, September and November the minimum. We will now try to figure out what could be the reason.

# 3 ) Identify periods with the strongest and weakest retention rates. What strategies could be implemented to improve retention during weaker months?

## Let us establish the definition of Retention Rate which we are going to follow. When Googled, we can see - $$Retention\ Rate\ (\%) = \left( \frac{\text{Number of customers from a cohort who bought in Month } X}{\text{Total unique customers who started in that Cohort}} \right) \times 100$$

We will follow the same strategy through out the EDA

### Calculate the Retention Matrix using Online_Sales Dataset

```python
Online_Sales['Transaction_Month'] = Online_Sales['Transaction_Date'].dt.to_period('M')
Online_Sales['Cohort_Month'] = Online_Sales.groupby('CustomerID')['Transaction_Month'].transform('min')

def get_month_difference(earliest_date, current_date):
    return (current_date.year - earliest_date.year) * 12 + (current_date.month - earliest_date.month)

Online_Sales['Cohort_Index'] = Online_Sales.apply(lambda row: get_month_difference(row['Cohort_Month'], row['Transaction_Month']), axis=1)
```

### Constructing the Retention Matrix

```python
# Group by Cohort_Month and Cohort_Index to count unique customers
cohort_counts = Online_Sales.groupby(['Cohort_Month', 'Cohort_Index']).agg(N_Customers=('CustomerID', 'nunique')).reset_index()
```

```python
# Calculate the initial cohort size (number of unique customers in the Cohort_Month)
cohort_sizes = cohort_counts[cohort_counts['Cohort_Index'] == 0][['Cohort_Month', 'N_Customers']]
cohort_sizes.rename(columns={'N_Customers': 'Cohort_Size'}, inplace=True)
```

```python
# Merge cohort counts with cohort sizes to calculate retention rate
retention_matrix_df = pd.merge(cohort_counts, cohort_sizes, on='Cohort_Month', how='left')
retention_matrix_df['Retention_Percentage'] = (retention_matrix_df['N_Customers'] / retention_matrix_df['Cohort_Size']) * 100
```

```python
# Pivot the table to create the retention matrix
retention_pivot = retention_matrix_df.pivot_table(index='Cohort_Month', columns='Cohort_Index', values='Retention_Percentage')
```

### Visualizing the Retention Matrix as a Heatmap

```python
plt.figure(figsize=(12, 8))
sns.heatmap(retention_pivot, annot=True, fmt='.1f', cmap='Blues', vmin=0, vmax=100)
plt.title('Customer Retention Matrix')
plt.xlabel('Cohort Index (Months Since First Purchase)')
plt.ylabel('Cohort Month')
plt.show()
```

##Conclusion:
> Strong Periods:



1.   Immediate Retention Rate (0,1 Index Cohort,means they are returning immediately) is high in June and July
2.   Long Term Retention Rate: Jan and Feb, there are good numbers even after Cohort Index is > 4.




> Weak Periods:


1.   Immediate Drop - Jan, Feb and October though for Jan it makes sense as it saw very high number of acquisition to begin with.
2.   Long Term Drop - September and October - Their immediate and long term both are bad.

# 10 ) Segment customers into groups such as Premium, Gold, Silver, and Standard. What targeted strategies can be developed for each segment to improve retention and revenue? (Use RFM segmentation techniques)

```python
snapshot_date = merged_sales_discounts['Transaction_Date'].max() + pd.Timedelta(days=1)
#Aggregate to create RFM values per customer
rfm_df = merged_sales_discounts.groupby('CustomerID').agg({
    'Transaction_Date': lambda x: (snapshot_date - x.max()).days, # Recency
    'Transaction_ID': 'nunique',                                  # Frequency
    'Net_Revenue': 'sum'                                          # Monetary
}).reset_index()

rfm_df.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']


# Compute RFM Scores (1 to 4 scale) using quantiles ---
# Recency: Lower is better (1 is oldest/worst, 4 is most recent/best)
rfm_df['R_Score'] = pd.qcut(rfm_df['Recency'], q=4, labels=[4, 3, 2, 1])

# Frequency & Monetary: Higher is better (4 is highest/best, 1 is lowest/worst)
# Use 'rank(method="first")' to safely handle heavily repeating low frequencies if necessary
rfm_df['F_Score'] = pd.qcut(rfm_df['Frequency'].rank(method='first'), q=4, labels=[1, 2, 3, 4])
rfm_df['M_Score'] = pd.qcut(rfm_df['Monetary'], q=4, labels=[1, 2, 3, 4])


# Map scores to tiers
# Create an aggregate metric to evaluate overall tier value
rfm_df['Total_RFM_Score'] = rfm_df['R_Score'].astype(int) + rfm_df['F_Score'].astype(int) + rfm_df['M_Score'].astype(int)

# Map the final scores into your 4 defined buckets
def assign_segment(score):
    if score >= 10:
        return 'Premium'
    elif score >= 8:
        return 'Gold'
    elif score >= 5:
        return 'Silver'
    else:
        return 'Standard'

rfm_df['Customer_Segment'] = rfm_df['Total_RFM_Score'].apply(assign_segment)

# Check your segment distributions
print(rfm_df['Customer_Segment'].value_counts())
```

## Conclusion - Customer Segment count for each category is visible above


---


Suggested Strategies:

**Premium** - Very high RFM score, we can start some Loyalty Programs like exclusive offers, early access to sales etc.

**Gold** - Good intent customers, we can encourage them for higher spend and make them Premium by sending them personal recommendations and offering them discounts on bigger expenses.

**Silver** - To increase their engagement, we can target promotions and coupons for them to come back often like 30 days cashback offer.

**Standard** - We can ask rollout some Welcome offer, some one time offers. We can also target them for feedback to understand how we can increase their engagement

# 14 ) Do customers who use coupons have a different average transaction value compared to those who do not?Conduct a statistical test to validate this hypothesis. What implications does this have for the company’s discount and coupon strategies?

```python
from scipy import stats

# Separate the Net_Revenue for transactions where coupons were used vs. not used
coupon_used_revenue = merged_sales_discounts[merged_sales_discounts['Coupon_Status'] == 'Used']['Net_Revenue'].dropna()
coupon_not_used_revenue = merged_sales_discounts[merged_sales_discounts['Coupon_Status'] == 'Not Used']['Net_Revenue'].dropna()

# Perform an independent samples t-test
# Null Hypothesis (H0): There is no significant difference in the average transaction value between customers who use coupons and those who do not.
# Alternative Hypothesis (H1): There is a significant difference in the average transaction value between customers who use coupons and those who do not.
t_statistic, p_value = stats.ttest_ind(coupon_used_revenue, coupon_not_used_revenue)

alpha = 0.05 # Significance level
if p_value < alpha:
    conclusion = "We reject the null hypothesis. There is a statistically significant difference in the average transaction value between customers who use coupons and those who do not."
else:
    conclusion = "We fail to reject the null hypothesis. There is no statistically significant difference in the average transaction value between customers who use coupons and those who do not."

print(f"\nConclusion: {conclusion}")
print(f"T-statistic: {t_statistic:.3f}, P-value: {p_value:.3f}")

# Further analysis of means
mean_used = coupon_used_revenue.mean()
mean_not_used = coupon_not_used_revenue.mean()

print(f"\nAverage Transaction Value (Coupon Used): ${mean_used:.2f}")
print(f"Average Transaction Value (Coupon Not Used): ${mean_not_used:.2f}")
```

## Conclusion: As per our test ATV is lesser for customers who used Coupon than who did not. Might mean that price sensitive customers do care about coupon. We can change are discount strategy, to have some minimum amount to avail coupon. Higher Spend for higher discounts to increase ATV.

```python
demographic_sales = pd.merge(merged_sales_discounts, CustomersData, on='CustomerID', how='left')

print("\n--- Analysis by Customer Location ---")

# Calculate average order value and order frequency per customer per location
location_purchase_behavior = demographic_sales.groupby(['CustomerID', 'Location']).agg(
    Avg_Order_Value=('Net_Revenue', 'mean'),
    Total_Order_Frequency=('Transaction_ID', 'nunique')
).reset_index()

# Group by Location to get overall averages for visualization
location_summary = location_purchase_behavior.groupby('Location').agg(
    Mean_Avg_Order_Value=('Avg_Order_Value', 'mean'),
    Mean_Order_Frequency=('Total_Order_Frequency', 'mean')
).reset_index().sort_values(by='Mean_Avg_Order_Value', ascending=False)

print("\nAverage Order Value and Frequency by Location:")
display(location_summary.round(2))

# Statistical Test (Kruskal-Wallis for non-normal distribution) for Avg_Order_Value
location_groups_value = [location_purchase_behavior['Avg_Order_Value'][location_purchase_behavior['Location'] == loc].dropna() for loc in location_summary['Location'].unique()]
if all(len(g) > 0 for g in location_groups_value):
    kw_h_value, kw_p_value = stats.kruskal(*location_groups_value)
    print(f"\nKruskal-Wallis Test for Average Order Value across Locations: H-statistic={kw_h_value:.3f}, P-value={kw_p_value:.3f}")
    if kw_p_value < 0.05:
        print("There is a statistically significant difference in average order value across different locations.")
    else:
        print("There is no statistically significant difference in average order value across different locations.")
else:
    print("Not enough data in one or more location groups for Kruskal-Wallis test (Average Order Value).")

# Statistical Test (Kruskal-Wallis for non-normal distribution) for Total_Order_Frequency
location_groups_frequency = [location_purchase_behavior['Total_Order_Frequency'][location_purchase_behavior['Location'] == loc].dropna() for loc in location_summary['Location'].unique()]
if all(len(g) > 0 for g in location_groups_frequency):
    kw_h_frequency, kw_p_frequency = stats.kruskal(*location_groups_frequency)
    print(f"\nKruskal-Wallis Test for Order Frequency across Locations: H-statistic={kw_h_frequency:.3f}, P-value={kw_p_frequency:.3f}")
    if kw_p_frequency < 0.05:
        print("There is a statistically significant difference in order frequency across different locations.")
    else:
        print("There is no statistically significant difference in order frequency across different locations.")
else:
    print("Not enough data in one or more location groups for Kruskal-Wallis test (Order Frequency).")



# --- 2. Analyze by Age Groups (using Tenure_Months) --- #
print("\n--- Analysis by Age Groups (Tenure_Months) ---")

# Define tenure bins dynamically based on actual data range
max_tenure = demographic_sales['Tenure_Months'].max()
tenure_bins = [0, 12, 24, 36, 60, max(60, max_tenure) + 1]
tenure_labels = ['0-12M', '13-24M', '25-36M', '37-60M', '60M+']
valid_bin_labels = [
    label for label, lo in zip(tenure_labels, tenure_bins[:-1])
    if lo <= max_tenure
]
demographic_sales['Tenure_Group'] = pd.cut(
    demographic_sales['Tenure_Months'], bins=tenure_bins, labels=tenure_labels, right=False
)

tenure_purchase_behavior = demographic_sales.groupby(['CustomerID', 'Tenure_Group'], observed=True).agg(
    Avg_Order_Value=('Net_Revenue', 'mean'),
    Total_Order_Frequency=('Transaction_ID', 'nunique')
).reset_index()

tenure_summary = tenure_purchase_behavior.groupby('Tenure_Group', observed=True).agg(
    Mean_Avg_Order_Value=('Avg_Order_Value', 'mean'),
    Mean_Order_Frequency=('Total_Order_Frequency', 'mean')
).reset_index().sort_values(by='Mean_Avg_Order_Value', ascending=False)

print("\nAverage Order Value and Frequency by Tenure Group:")
display(tenure_summary.round(2))

def run_kruskal(df, group_col, value_col, label):
    """Robust Kruskal-Wallis: drops empty groups instead of failing entirely."""
    groups = [
        df[value_col][df[group_col] == g].dropna()
        for g in df[group_col].dropna().unique()
    ]
    valid_groups = [g for g in groups if len(g) > 0]
    if len(valid_groups) >= 2:
        h, p = stats.kruskal(*valid_groups)
        print(f"\nKruskal-Wallis Test for {label}: H-statistic={h:.3f}, P-value={p:.3f}")
        if p < 0.05:
            print(f"There is a statistically significant difference in {label.lower()}.")
        else:
            print(f"There is no statistically significant difference in {label.lower()}.")
    else:
        print(f"Not enough non-empty groups to run Kruskal-Wallis test for {label}.")

run_kruskal(tenure_purchase_behavior, 'Tenure_Group', 'Avg_Order_Value',
            'Average Order Value across Tenure Groups')
run_kruskal(tenure_purchase_behavior, 'Tenure_Group', 'Total_Order_Frequency',
            'Order Frequency across Tenure Groups')


# --- 3. Analyze by Delivery Charge Tiers --- #
print("\n--- Analysis by Delivery Charge Tiers ---")

delivery_charge_bins = pd.cut(
    demographic_sales['Delivery_Charges'], bins=4, labels=['Tier 1', 'Tier 2', 'Tier 3', 'Tier 4']
)
if demographic_sales['Delivery_Charges'].isna().any():
    delivery_charge_bins = delivery_charge_bins.cat.add_categories('Unknown').fillna('Unknown')
demographic_sales['Delivery_Charge_Tier'] = delivery_charge_bins

delivery_purchase_behavior = demographic_sales.groupby(['CustomerID', 'Delivery_Charge_Tier'], observed=True).agg(
    Avg_Order_Value=('Net_Revenue', 'mean'),
    Total_Order_Frequency=('Transaction_ID', 'nunique')
).reset_index()

delivery_summary = delivery_purchase_behavior.groupby('Delivery_Charge_Tier', observed=True).agg(
    Mean_Avg_Order_Value=('Avg_Order_Value', 'mean'),
    Mean_Order_Frequency=('Total_Order_Frequency', 'mean')
).reset_index().sort_values(by='Mean_Avg_Order_Value', ascending=False)

print("\nAverage Order Value and Frequency by Delivery Charge Tier:")
display(delivery_summary.round(2))

run_kruskal(delivery_purchase_behavior, 'Delivery_Charge_Tier', 'Avg_Order_Value',
            'Average Order Value across Delivery Charge Tiers')
run_kruskal(delivery_purchase_behavior, 'Delivery_Charge_Tier', 'Total_Order_Frequency',
            'Order Frequency across Delivery Charge Tiers')
```

## Conclusion:


1.   Analysis by Customer Location : Since P>0.05, statistically there is no difference in how peoply buy across locations.
2.   Analysis by Age Groups : Same, P>0.05, Age Group doesn't impact Average Order Value or Frequency.
3. Analysis by Delivery Charge Tiers: P=0, Delivery Charge certainly impacts our Order Value and Frequency.

Insights for Marketing and Strategy: Frequency for Tier1 is high but Order Value is smaller, so we can implement some Minimum Cart Amount for Shipping to be Free. # 18 ) Evaluate how taxes and delivery charges influence customer spending behavior. Are there opportunities to adjust pricing strategies to improve customer satisfaction and revenue?

```python
# Collective Cost Calculation
sales_with_tax = pd.merge(merged_sales_discounts, Tax_amount, on='Product_Category', how='left')
sales_with_tax['Tax_Amount_Item'] = sales_with_tax['Net_Revenue'] * sales_with_tax['GST']

# Aggregate directly to transaction-level total bills
bills_df = sales_with_tax.groupby(['Transaction_ID', 'Delivery_Charges']).agg(
    Net_Revenue=('Net_Revenue', 'sum'),
    Tax_Amount=('Tax_Amount_Item', 'sum')
).reset_index()

# The collective impact metric (The total bill)
bills_df['Total_Bill_Amount'] = bills_df['Net_Revenue'] + bills_df['Tax_Amount'] + bills_df['Delivery_Charges']

# Bin into Delivery Charge Groups
bills_df['Delivery_Group'] = pd.cut(
    bills_df['Delivery_Charges'], bins=4, labels=['Tier 1', 'Tier 2', 'Tier 3', 'Tier 4']
).cat.add_categories('Unknown').fillna('Unknown')


#ANOVA Test
print("### ANOVA Test: Total Bill Amounts Across Delivery Charge Groups")

# Isolate total bill amounts for each group
group_bills = [bills_df['Total_Bill_Amount'][bills_df['Delivery_Group'] == tier].dropna()
               for tier in ['Tier 1', 'Tier 2', 'Tier 3', 'Tier 4']]

# Run the test
f_stat, p_val = stats.f_oneway(*[g for g in group_bills if len(g) > 0])
print(f"ANOVA F-statistic: {f_stat:.3f}, P-value: {p_val:.3f}")

if p_val < 0.05:
    print("Conclusion: Delivery charge groupings structurally and significantly impact the overall customer bill amount.")
else:
    print("Conclusion: Overall bills do not vary significantly across delivery charge groups.")
```

## Conclusion: The ANOVA shows that delivery charge tiers certainly influence total consumer bill amounts ($F=165.478, p=0.000$). Customers are trying to make cart value high to psychologically justify delivery fee. We can feed this result and result from Q15 to marketing team to make our case strong for conclusion of Q15.
